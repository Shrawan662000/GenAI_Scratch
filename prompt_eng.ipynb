### SDK Comparison
#### Old
client.chat.completions.create(...)
#### New
client.responses.create(...)

| Chat Completions             | Responses API                                     |
| ---------------------------- | ------------------------------------------------- |
| `messages`                   | `input`                                           |
| `system` message             | `instructions` (preferred) or `system` in `input` |
| `max_tokens`                 | `max_output_tokens`                               |
| `temperature`                | Not supported for GPT-5 models                    |
| `top_p`                      | Not supported for GPT-5 models                    |
| `functions`                  | `tools`                                           |
| `function_call`              | `tool_choice`                                     |
| `choices[0].message.content` | `output_text`                                     |


```python
from openai import OpenAI
import os
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

response = client.responses.create(
    model="gpt-5.5",
    input=[
        {
            "role": "system",
            "content": (
                "You are a professional multilingual translator. "
                "Your task is to translate any input text into Hindi.\n\n"
                "Rules:\n"
                "1. Detect the source language automatically.\n"
                "2. Translate ONLY into Hindi.\n"
                "3. Preserve meaning, tone, formatting, numbers, URLs, emails, and code.\n"
                "4. Translate idioms naturally.\n"
                "5. If the input is already in Hindi, return it unchanged.\n"
                "6. Return only the translated Hindi text."
            )
        },
        {
            "role": "assistant",
            "content": (
                "Understood. I will translate any input into Hindi while "
                "preserving the original meaning and formatting. "
                "I will return only the translated Hindi text."
            )
        },
        {
            "role": "user",
            "content": "Artificial Intelligence is transforming the world."
        }
    ],
    max_output_tokens=500,
    store=False,
)

print(response.output_text)
```

```python
#5. Structured JSON Output
import json

response = client.responses.create(
    model="gpt-5.5",
    input="John is 30 years old and works at Google.",
    text={
        "format": {
            "type": "json_schema",  # different from the default "text" format
            "name": "person",
            "schema": {
                "type": "object",
                "properties": {
                    "name": {"type": "string"},
                    "age": {"type": "integer"},
                    "company": {"type": "string"}
                },
                "required": [
                    "name",
                    "age",
                    "company"
                ],
                "additionalProperties": False
            }
        }
    }
)

print(json.loads(response.output_text))
```

```python
#15. Conversation Continuation
first = client.responses.create(
    model="gpt-5.5",
    input="What is AI?"
)

second = client.responses.create(
    model="gpt-5.5",
    previous_response_id=first.id,
    input="Explain with an example."
)

print(second.output_text)
```

### Complete client.responses.create() Parameters (Most Common)
| Parameter              | Type           | Supported in GPT-5.5           | Purpose                                                                  | Example                                         |
| ---------------------- | -------------- | ------------------------------ | ------------------------------------------------------------------------ | ----------------------------------------------- |
| `model`                | `str`          | ✅                              | Model to use                                                             | `"gpt-5.5"`                                     |
| `input`                | `str` | `list` | ✅                              | User prompt or role-based conversation                                   | `"Hello"` or `[{"role":"user","content":"Hi"}]` |
| `instructions`         | `str`          | ✅                              | High-level behavior or system instructions                               | `"You are a helpful tutor."`                    |
| `tools`                | `list`         | ✅                              | Built-in tools or custom functions                                       | `web_search`, `file_search`, `function`         |
| `tool_choice`          | `str` | `dict` | ✅                              | Control whether or which tool is used                                    | `"auto"`                                        |
| `previous_response_id` | `str`          | ✅                              | Continue a previous conversation                                         | `"resp_abc123"`                                 |
| `reasoning`            | `dict`         | ✅ (Reasoning models)           | Configure reasoning effort                                               | `{"effort":"high"}`                             |
| `text`                 | `dict`         | ✅                              | Output options such as verbosity or JSON Schema                          | `{"verbosity":"high"}`                          |
| `max_output_tokens`    | `int`          | ✅                              | Maximum number of output tokens                                          | `1000`                                          |
| `metadata`             | `dict`         | ✅                              | Attach custom metadata                                                   | `{"project":"GenAI"}`                           |
| `store`                | `bool`         | ✅                              | Store the response for later retrieval                                   | `False`                                         |
| `stream`               | `bool`         | ✅                              | Stream the response incrementally                                        | `True`                                          |
| `temperature`          | `float`        | ❌ **Not supported in GPT-5.5** | Controls randomness (lower = more deterministic, higher = more creative) | `0.7`                                           |
| `top_p`                | `float`        | ❌ **Not supported in GPT-5.5** | Nucleus sampling; alternative to temperature                             | `0.9`                                           |
| `frequency_penalty`    | `float`        | ❌ **Not supported in GPT-5.5** | Reduces repeated words or phrases                                        | `0.5`                                           |
| `presence_penalty`     | `float`        | ❌ **Not supported in GPT-5.5** | Encourages introducing new topics                                        | `0.5`                                           |
| `user`*                | `str`          | Depends on API/model           | End-user identifier for monitoring and abuse detection                   | `"user123"`                                     |


### Notice:

instructions → defines how the model should behave.

input → contains the actual user conversation or multimodal input.

| Scenario                             | Recommendation                                     |
| ------------------------------------ | -------------------------------------------------- |
| New project                          | ✅ Use `instructions`                               |
| Migrating from Chat Completions      | ✅ Keep the existing `system` message if convenient |
| Need to maintain old message history | ✅ Use `system` in `input`                          |
| Use both for the same purpose        | ❌ Generally avoid                                  |


#### Rule of thumb
instructions → Define the assistant's overall behavior and rules.

input → Contains the conversation (user messages, assistant history, and, if needed for compatibility, system messages).
