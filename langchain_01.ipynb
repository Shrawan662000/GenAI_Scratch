#### batch()

Suppose you have 100 prompts.

Without batch:

for question in questions:

    llm.invoke(question)

#### With batch:



````python
questions = [
    "What is AI?",
    "What is ML?",
    "What is Deep Learning?",
]

responses = llm.batch(questions)

for response in responses:
    print(response.content)
````

#### stream()

Streaming returns tokens incrementally instead of waiting for the complete answer.



````python
for chunk in llm.stream(
    "Explain LangChain."
):
    print(chunk.content, end="", flush=True)
````

#### invoke() vs batch() vs stream()
| Method     | Use Case                    | Returns                    |
| ---------- | --------------------------- | -------------------------- |
| `invoke()` | One input                   | `AIMessage`                |
| `batch()`  | Multiple independent inputs | List of `AIMessage`        |
| `stream()` | Incremental output          | Iterator of message chunks |


#### Streaming Example

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# Create LLM
llm = ChatOpenAI(
    model="gpt-5.4-mini",
    temperature=0,
    streaming=True,
    stream_usage=True
)

# User prompt
messages = [
    HumanMessage(content="Explain Retrieval-Augmented Generation (RAG) in simple words.")
]

print("Response:\n")

# Stream response token by token
for chunk in llm.stream(messages):
    if chunk.content:
        print(chunk.content, end="", flush=True)

print("\n\nStreaming completed.")
````

#### Streaming with Token Usage

````python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(
    model="gpt-5.4-mini",
    temperature=0,
    streaming=True,
    stream_usage=True
)

messages = [
    HumanMessage(content="What is Generative AI?")
]

final_response = None

print("Response:\n")

for chunk in llm.stream(messages):

    if chunk.content:
        print(chunk.content, end="", flush=True)

    # Save last chunk (contains metadata)
    final_response = chunk

print("\n")

# Token Usage
if final_response:
    print("Usage:")
    print(final_response.response_metadata.get("token_usage"))
````

#### 1. Tuple Format (Shorthand)

```python
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "I love programming."),
]
```

This is a simplified shorthand supported by LangChain. Internally, LangChain automatically converts it into message objects.

#### Equivalent to:

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(
        content="You are a helpful assistant that translates English to French. Translate the user sentence."
    ),
    HumanMessage(content="I love programming.")
]
```

#### Which one should you use?
Shorthand tuples: Great for quick examples and prototypes.

Message objects (HumanMessage, SystemMessage, AIMessage): Preferred in production because they're more explicit, support richer metadata, and are easier to work with in larger applications.
